# Dask dashboard quick check

Use this notebook to spin up a small local Dask cluster and view the live dashboard (tasks, CPU, memory). The cluster is lightweight by default; tweak worker counts/memory as needed for your machine.

```python
from dask.distributed import Client, LocalCluster

# Start a small local cluster; dashboard on port 8787
cluster = LocalCluster(
    n_workers=2,
    threads_per_worker=2,
    memory_limit="2GB",
    dashboard_address=":8787",
)
client = Client(cluster)
client
```

### Accessing the dashboard
- Local: open the `dashboard_link` above (usually http://localhost:8787/status).
- Remote/SSH: `ssh -L 8787:localhost:8787 <user>@<host>` then open http://localhost:8787/status in your browser.

```python
from dask import delayed
import time

# A tiny workload to make the task stream/update graphs interesting
def sleep_add(x, y, delay=0.5):
    time.sleep(delay)
    return x + y

futures = [delayed(sleep_add)(i, i) for i in range(12)]
total = delayed(sum)(futures)
result = total.compute()
result
```

When you are done, shut down the cluster to free resources:

```python
client.close()
cluster.close()
```
